OANDA DOCUMENT: https://developer.oanda.com/rest-live-v20/instrument-df/

```python
import requests
import defs
import pandas as pd
```

```python
session = requests.Session()
```

we want to maintian "persistant connection" - using a session, to prevent creating a new request creating server 

```python
url = f"{defs.OANDA_URL}/accounts/{defs.ACCOUNT_ID}/instruments"
```

```python
url
```

```python
response = session.get(url, params=None, headers=defs.SECURE_HEADER)
```

```python
response.status_code
```

```python
data = response.json()
```

```python
data.keys()
```

```python
instruments = data['instruments']
```

```python
instruments[0].keys()
```

```python
instrument_data = []
for item in instruments:
    new_obj = dict(
        name = item["name"],
        type = item["type"],
        displayName = item["displayName"],
        pipLocation = item["pipLocation"],
        marginRate = item["marginRate"]
    )
    instrument_data.append(new_obj)
```

```python
for item in instrument_data[0:3]:
    print(item)
```

```python
instrument_df = pd.DataFrame.from_dict(instrument_data)
```

```python
instrument_df
```

```python
instrument_df.to_pickle("instruments.pkl")
```
